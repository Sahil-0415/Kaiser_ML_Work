```python
name = "Sahil"
first_char = name[0]
length = len(name)
middle = int(length/2)
middle_char=name[middle]
last_char = name[-1]

print(first_char+middle_char+last_char)
```

```python
name = "SahilSaiSahil"
length = len(name)
middle = int(length/2)

mid = name[middle]
mid_p = name[middle-1]
mid_f = name[middle+1]

print(mid_p+mid+mid_f)
```
