```python

fruits = ["apple", "banana", "cherry"]
print(f"Original list: {fruits}")

# 2. append() - Adds an element to the end of the list
fruits.append("orange")
print(f"After append(): {fruits}")

# 3. extend() - Adds all elements of another list (or iterable) to the end
tropical = ["mango", "pineapple"]
fruits.extend(tropical)
print(f"After extend(): {fruits}")


fruits.insert(1, "blueberry")
print(f"After insert(): {fruits}")

fruits.remove("banana")
print(f"After remove(): {fruits}")


removed_item = fruits.pop(2)  # Removes item at index 2
last_item = fruits.pop()      # Removes the last item
print(f"After pop(2) [removed '{removed_item}'] and pop() [removed '{last_item}']: {fruits}")

fruits.extend(["apple", "grapes"])
print(f"Updated list for searches: {fruits}")
apple_index = fruits.index("apple")
print(f"The index() of the first 'apple' is: {apple_index}")


apple_count = fruits.count("apple")
print(f"The count() of 'apple' in the list is: {apple_count}")


fruits.reverse()
print(f"After reverse(): {fruits}")

fruits.sort()
print(f"After sort(): {fruits}")

fruits_backup = fruits.copy()
print(f"The copy() stored in fruits_backup: {fruits_backup}")


fruits.clear()
print(f"After clear(): {fruits}")
print(f"Verified backup is still intact: {fruits_backup}")
```

```python
# Count Even and Odd Numbers
name = [1, 5, 4, 5]
even = 0
odd = 0
for i in name:
    if i % 2 == 0:
        even = even+1;
    else:
        odd = odd+1
print(even)
print(odd)            
```
