```python
import pandas as pd
```

```python
df = pd.read_csv("dirty_data_ship.csv")
df.columns
```

```python
s1 = df.melt(id_vars=['Order ID'],var_name='Segment',value_name='Sales')
```

```python
s1
```

```python
s1 = s1.dropna(subset=["Sales"])
```

```python
s1['Segment'] = (
    s1['Segment']
    .str.replace('_', ' ')
    .str.title()
)
```

```python
s1 = s1.rename(
    columns={
        'Order ID':'OrderID'
    }
)
```

```python
print(s1.head(20))
```

```python
s1.to_csv("Clean_Ship_data.csv" , index = False)
```
