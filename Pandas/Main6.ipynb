```python
import pandas as pd
df = pd.read_csv("titanic.csv")
df["Age_status"] = df["Age"].isna().map({True : "Missing" , False:"Known"})
```

```python
df.head(20)
```

```python
print("=" * 65)
print("       TITANIC EDA: SYSTEMIC MISSING AGE INVESTIGATION       ")
print("=" * 65)
```

```python
sm = df.groupby("Age_status")["Survived"].agg(["count","mean"])
```

```python
sm
```

```python
sm["mean"] = (sm["mean"] * 100).round(2).astype(str) + "%"
sm.columns = ["Total Passengers", "Survival Rate"]
print(sm)
```

```python
pclass_crosstab = pd.crosstab(df["Pclass"], df["Age_status"])
pclass_pct = pd.crosstab(df["Pclass"], df["Age_status"], normalize="columns").mul(100).round(2).astype(str) + "%"
pclass_summary = pd.concat([pclass_crosstab, pclass_pct], axis=1, keys=["Counts", "Proportion within Group"])
print(pclass_summary)
```

```python
fare_summary = df.groupby("Age_status")["Fare"].describe()[["mean", "25%", "50%", "75%", "max"]].round(2)
fare_summary.columns = ["Mean Fare", "25th Pct", "Median (50th)", "75th Pct", "Max Fare"]
print(fare_summary)
```

```python
port_crosstab = pd.crosstab(df["Embarked"], df["Age_status"])
port_missing_pct = pd.crosstab(df["Embarked"], df["Age_status"], normalize="index").mul(100).round(2).astype(str) + "%"
port_summary = pd.concat([port_crosstab, port_missing_pct], axis=1, keys=["Counts", "Proportion within Port"])
print(port_summary)
```

```python
import pandas as pd
df1 = pd.read_csv("titanic.csv")
df1["Fam_size"] = df1["SibSp"]+df1["Parch"]
df1["Fam_size"].describe()
```

```python
import pandas as pd
df2 = pd.read_csv("titanic.csv")

q1 = df2["Fare"].quantile(0.25)
q2 = df2["Fare"].quantile(0.75)

Iqr = q2-q1
lower = q1 - 1.5 * Iqr
higher = q2 + 1.5 * Iqr

outlier = (df2["Fare"] < lower) | (df2["Fare"] > higher)
df2["Survied1"] = outlier
print(outlier)
df2.info()
df2["Survied1"].info()


```

```python
ss = df2.groupby("Survied1")["Survived"].sum()
ss
```

```python
print(pd.crosstab(df2["Survied1"] , df2["Survived"]))
```
