```python
import pandas as pd
s2 = pd.read_csv("my_file1.csv")
s2 = s2.fillna(0)
s2["Actual gross"] = s2["Actual gross"].str.replace(',', '', regex=False).str.replace('$' , "" ,regex=False)
s2["Actual gross"] = s2["Actual gross"].str.replace('[e]', '', regex=False)
s2["Actual gross"] = s2["Actual gross"].str.replace('[b]', '', regex=False)
s2["Actual gross(in 2022 dollars)"] = s2["Actual gross(in 2022 dollars)"].str.replace(',', '', regex=False).str.replace('$' , "" ,regex=False)
s2["Average gross"] = s2["Average gross"].str.replace(',', '', regex=False).str.replace('$' , "" ,regex=False)
s2["Peak"] = s2["Peak"].str.replace(r"\[\d+\]","", regex=True)
s2["All Time Peak"] = s2["All Time Peak"].str.replace(r"\[\d+\]","", regex=True)
s2
```

```python
s2["Peak"] = s2["Peak"].str.extract(r'(\d+)')

s2.head(20)
```

```python
s2['Tour title'] = (

s2['Tour title']

.str.replace(r'\[.*?\]', '', regex=True) # remove [4], [a], [10]

.str.replace('†', '', regex=False) # remove dagger symbol

.str.replace('‡', '', regex=False) # remove double dagger

.str.strip() # remove extra spaces

)
s2
```

```python
s2["Artist"] = s2["Artist"].str.replace("" , "" , regex=False)
s2
```

```python
s2["Year(s)"] = s2["Year(s)"].str.extract(r'(\d{4})')
s2
```

```python
s2["Peak"] = s2["Peak"].str.extract(r'(\d+)')
s2["Peak"] = s2["Peak"].fillna(0 , inplace=True)
s2
```

```python
s2["Actual gross(in 2022 dollars)"] = (
    s2["Actual gross(in 2022 dollars)"].fillna(0 , inplace=True)
)
s2["Actual gross"].dtype
```

```python
s2["Actual gross"] = pd.to_numeric(
    s2["Actual gross"]
    .astype(str)
    .str.replace(r'[^0-9.]', '', regex=True),
    errors='coerce'
)
```

```python
s2["Average gross"] = pd.to_numeric(
    s2["Average gross"]
    .astype(str)
    .str.replace(r'[^0-9.]', '', regex=True),
    errors='coerce'
)
```
