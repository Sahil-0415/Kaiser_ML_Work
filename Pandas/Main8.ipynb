```python
import pandas as pd
import regex as re
dff = pd.read_csv("healthcare_dataset.csv")
dff.describe()
dff["Medical Condition"].nunique()
```

```python

dff["Name"] = dff["Name"].str.lower()
dff["Name"]= dff["Name"].str.capitalize()
dff["Name"]= dff["Name"].str.title()
dff["Name"]= dff["Name"].str.lstrip(" ")
dff["Name"] = dff["Name"].str.strip()
max_len = dff["Name"].str.len().max()
dff["Name"] = dff["Name"].str.ljust(max_len)


dff["Name"].head(20)
```

```python
prefixes = ("Dr. ","MD. ", "DDS. ","Ms. ","Mrs. ","Mr. ")

for p in prefixes:
    dff["Name"] = dff["Name"].str.removeprefix(p)

```

```python
max_len = dff["Doctor"].str.len().max()
dff["Doctor"] = dff["Doctor"].str.rjust(max_len)
dff["Doctor"] = dff["Doctor"].str.lower()


pattern = r'[\s,]+(ph\.?d\.?|m\.?d\.?)\.?$'
dff['Doctor'] = dff['Doctor'].str.replace(pattern, '', regex=True, flags=re.IGNORECASE).str.strip()

prefixes = ("dr. ","md ", "dds. ","ms. ","mrs. ","mr. ","dvm. ","jr. ")

for p in prefixes:
    dff["Doctor"] = dff["Doctor"].str.removeprefix(p)

max_len = dff["Doctor"].str.len().max()
dff["Doctor"] = dff["Doctor"].str.ljust(max_len)
dff["Doctor"]= dff["Doctor"].str.capitalize()
dff["Doctor"]= dff["Doctor"].str.title()
 

dff["Doctor"].head(60)
```

```python
rare_columns = [
    "Blood Type", "Medical Condition", "Insurance Provider",
    "Medication", "Admission Type"
]

for col in rare_columns:
    print("\n================", col, "================")
    percentages = dff[col].value_counts(normalize=True) * 100
    print(percentages)

    for threshold in [1, 2, 5]:
        print(f"Below {threshold}%:")
        print(percentages[percentages < threshold])

# Function for modelling: group rare levels
def group_rare_categories(data, column, threshold=0.05):
    proportions = data[column].value_counts(normalize=True)
    rare = proportions[proportions < threshold].index
    data[column] = data[column].replace(rare, "Other")
    return data

# Example:
dff = group_rare_categories(dff, "Insurance Provider", 0.05)
```

```python
# IQR-based global billing anomalies
Q1 = dff["Billing Amount"].quantile(0.25)
Q3 = dff["Billing Amount"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

dff["Billing Anomaly"] = (
    (dff["Billing Amount"] < lower) |
    (dff["Billing Amount"] > upper)
)

print("Number of billing anomalies:", dff["Billing Anomaly"].sum())

# More business-relevant: compare cost to condition-specific median
condition_median = dff.groupby("Medical Condition")["Billing Amount"].transform("median")

dff["Cost_vs_Condition_Median"] = (
    dff["Billing Amount"] / condition_median
)

business_anomalies = dff[dff["Cost_vs_Condition_Median"] > 2]

print(business_anomalies[
    ["Name", "Medical Condition", "Billing Amount",
     "Cost_vs_Condition_Median"]
].head(20))
```
