```python
import pandas as pd
```

```python
df = pd.read_csv("healthcare_dataset.csv")
```

```python
Q1 = df['Age'].quantile(0.25)
Q3 = df['Age'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = df[(df['Age'] < lower_bound) | (df['Age'] > upper_bound)]
print(f"Lower Threshold Boundary: {lower_bound}")
print(f"Upper Threshold Boundary: {upper_bound}")
print("\nDetected Outliers in 'age' column:")
print(outliers)
```

```python
Q1 = df['Billing Amount'].quantile(0.25)
Q3 = df['Billing Amount'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 0.5 * IQR
upper_bound = Q3 + 0.5 * IQR
outliers = df[(df['Billing Amount'] < lower_bound) | (df['Billing Amount'] > upper_bound)]
print(f"Lower Threshold Boundary: {lower_bound}")
print(f"Upper Threshold Boundary: {upper_bound}")
print("\nDetected Outliers in 'Billing Amount' column:")
print(len(outliers))
```

```python
import pandas as pd


mean_billing = df['Billing Amount'].mean()
std_billing = df['Billing Amount'].std()
df['Z_Score'] = (df['Billing Amount'] - mean_billing) / std_billing
outliers = df[df['Z_Score'].abs() > 1.9]
print(f"Dataset Mean: {mean_billing}")
print(f"Dataset Standard Deviation: {std_billing}\n")
print(f"Found {len(outliers)} outliers using the Z-score method:")
print(outliers[['Name', 'Age', 'Billing Amount', 'Z_Score']])
```
