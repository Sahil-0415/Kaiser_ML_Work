```python
import pandas as pd
df = pd.read_excel("Product-Sales-Region.xlsx")
df.head(5)
```

```python
df.info()
```

```python
import pandas as pd
import numpy as np

# Select numerical columns
num_cols = df.select_dtypes(include=['int64', 'float64']).columns

# Store results
results = []

for col in num_cols:

    mean_val = df[col].mean()
    median_val = df[col].median()
    mode_val = df[col].mode()[0]

    skew_val = df[col].skew()
    kurt_val = df[col].kurt()

    # Determine skewness type
    if skew_val > 0:
        skew_type = "Positively Skewed (Right Tail)"
    elif skew_val < 0:
        skew_type = "Negatively Skewed (Left Tail)"
    else:
        skew_type = "Symmetric"

    # Determine kurtosis type
    if kurt_val > 0:
        kurt_type = "Leptokurtic"
    elif kurt_val < 0:
        kurt_type = "Platykurtic"
    else:
        kurt_type = "Mesokurtic"

    results.append([
        col,
        round(mean_val, 2),
        round(median_val, 2),
        round(mode_val, 2),
        round(skew_val, 2),
        skew_type,
        round(kurt_val, 2),
        kurt_type
    ])

# Create summary dataframe
eda_summary = pd.DataFrame(results,
                           columns=[
                               "Column",
                               "Mean",
                               "Median",
                               "Mode",
                               "Skewness",
                               "Skew Type",
                               "Kurtosis",
                               "Kurtosis Type"
                           ])

print(eda_summary)
```
