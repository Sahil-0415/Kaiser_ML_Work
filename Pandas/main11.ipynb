```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load data
df = pd.read_csv('Clean_Csv_File.csv')

# Correlation matrix
corr_matrix = df.corr(numeric_only=True)

# Heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap')
plt.show()
```

```python
import pandas as pd
df = pd.read_csv("houseprice-dirty.csv")
df.describe()
```

```python
columns_to_corr = ['PoolArea', 'MiscVal', 'MoSold', 'YrSold',"SalePrice"]
corr_matrix = df[columns_to_corr].corr()

# 3. Set up the matplotlib figure
plt.figure(figsize=(8, 6))

# 4. Generate the heatmap
sns.heatmap(
    corr_matrix, 
    annot=True,          # Displays the correlation numbers inside the squares
    cmap='coolwarm',     # Color gradient (blue for negative, red for positive)
    fmt=".2f",           # Rounds numbers to 2 decimal places
    vmin=-1, vmax=1      # Fixes the color scale bounds between -1 and 1
)

plt.title('Correlation Matrix with SalePrice')
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load your dataset
df = pd.read_excel("Product-Sales-Region.xlsx")

# Create pair plot
dff = df[["Quantity" , "Discount" , "TotalPrice","RegionManager"]]
sns.pairplot(dff, hue="RegionManager",diag_kind="hist")

# Display plot
plt.show()
```
