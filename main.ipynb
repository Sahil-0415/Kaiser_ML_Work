```python
import time
import numpy as np

# Loop - based operation
my_list = list(range(1000000))
start_time_loop = time.time()
result_loop = []
for num in my_list:
    result_loop.append(num * 2)
end_time_loop = time.time()
loop_time = end_time_loop - start_time_loop

# Vectorized operation
my_array = np.array(my_list)
start_time_vectorized = time.time()
result_vectorized = my_array * 2
end_time_vectorized = time.time()
vectorized_time = end_time_vectorized - start_time_vectorized

print(f"Time taken by loop: {loop_time} seconds")
print(f"Time taken by vectorized operation: {vectorized_time} seconds")
```

```python
import pandas as pd

df = pd.read_excel("Online-Store-Orders.xlsx")

target = 3000

for start in range(len(df)):
    current_sum = 0
    window = []

    for end in range(start, len(df)):
        value = df["TotalPrice"].iloc[end]

        current_sum += value
        window.append(value)

        if current_sum >= target:
            print(f"\nWindow starting at row {start}")

            print("Values:", window)
            print("Window Size:", len(window))
            print("Sum:", current_sum)

            break
```
