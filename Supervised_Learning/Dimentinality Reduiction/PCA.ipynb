```python
import numpy as np

X = np.array([
    [2.5, 2.4],
    [0.5, 0.7],
    [2.2, 2.9],
    [1.9, 2.2],
    [3.1, 3.0],
    [2.3, 2.7],
    [2.0, 1.6],
    [1.0, 1.1],
    [1.5, 1.6],
    [1.1, 0.9]
])

print(X.shape)
```

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Standardize
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# PCA
pca = PCA(n_components=1)

X_pca = pca.fit_transform(X_scaled)

print(X_pca)
```

```python
X_pca.shape
```

```python
print(pca.explained_variance_ratio_)
```

```python
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

X_scaled = StandardScaler().fit_transform(X)

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

plt.figure(figsize=(6, 5))
plt.scatter(X_scaled[:, 0], X_scaled[:, 1])

plt.quiver(
    0, 0,
    pca.components_[0, 0],
    pca.components_[0, 1],
    color='red',
    scale=2,
    label='PC1'
)

plt.quiver(
    0, 0,
    pca.components_[1, 0],
    pca.components_[1, 1],
    color='blue',
    scale=2,
    label='PC2'
)

plt.legend()
plt.title("Principal Components")
plt.show()
```

```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

df = pd.read_csv("Country-data.csv")
df = df.drop("country" , axis=1)
print(df.shape)

scaler = StandardScaler()
X_scaledd = scaler.fit_transform(df)

pca = PCA(n_components=3)
x_pcaa = pca.fit_transform(X_scaledd)
print(x_pcaa.shape)
```
