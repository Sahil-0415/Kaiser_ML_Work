```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score, classification_report

# =========================
# Load Dataset
# =========================

df = pd.read_csv("Credit.csv")

# =========================
# Feature Scaling
# =========================

scaler = StandardScaler()

df[["RevolvingUtiOfUnsecuredLines", "DebtRatio"]] = scaler.fit_transform(
    df[["RevolvingUtiOfUnsecuredLines", "DebtRatio"]]
)

# =========================
# Encode Categorical Columns
# =========================

label = LabelEncoder()

# Example:
# df["Gender"] = label.fit_transform(df["Gender"])

# =========================
# Features and Target
# =========================

X = df.drop("Credit", axis=1)
y = df["Credit"]

# =========================
# Train Test Split
# =========================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

# =========================
# Find Best Alpha (Post Pruning)
# =========================

temp_tree = DecisionTreeClassifier(
    random_state=42,
    class_weight="balanced"
)

path = temp_tree.cost_complexity_pruning_path(X_train, y_train)

ccp_alphas = path.ccp_alphas

trees = []
train_scores = []
test_scores = []

for alpha in ccp_alphas:

    clf = DecisionTreeClassifier(
        criterion="gini",
        max_depth=5,            # Pre-pruning
        min_samples_split=20,   # Pre-pruning
        min_samples_leaf=10,    # Pre-pruning
        class_weight="balanced",
        random_state=42,
        ccp_alpha=alpha         # Post-pruning
    )

    clf.fit(X_train, y_train)

    trees.append(clf)

    train_scores.append(clf.score(X_train, y_train))
    test_scores.append(clf.score(X_test, y_test))

# Best alpha
best_alpha = ccp_alphas[test_scores.index(max(test_scores))]

print("Best CCP Alpha:", best_alpha)

# =========================
# Final Model
# =========================

model = DecisionTreeClassifier(
    criterion="gini",
    max_depth=5,
    min_samples_split=20,
    min_samples_leaf=10,
    class_weight="balanced",
    random_state=42,
    ccp_alpha=best_alpha
)

model.fit(X_train, y_train)

# =========================
# Prediction
# =========================

y_pred = model.predict(X_test)

# =========================
# Evaluation
# =========================

print("\nAccuracy:", accuracy_score(y_test, y_pred))

print("\nClassification Report\n")
print(classification_report(y_test, y_pred))

# =========================
# Visualize Accuracy vs Alpha
# =========================

plt.figure(figsize=(10,6))

plt.plot(
    ccp_alphas,
    train_scores,
    marker="o",
    label="Train Accuracy"
)

plt.plot(
    ccp_alphas,
    test_scores,
    marker="o",
    label="Test Accuracy"
)

plt.xlabel("CCP Alpha")
plt.ylabel("Accuracy")
plt.title("Accuracy vs CCP Alpha")
plt.legend()

plt.show()

# =========================
# Plot Final Tree
# =========================

plt.figure(figsize=(25,12))

plot_tree(
    model,
    feature_names=X.columns,
    class_names=[str(c) for c in model.classes_],
    filled=True,
    rounded=True,
    fontsize=8
)

plt.show()
```
