```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.ensemble import AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report, accuracy_score

# 1. Load the Iris dataset
iris = load_iris()
X, y = iris.data, iris.target

# 2. Split the dataset (70% train, 30% test)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 3. Initialize AdaBoost 
# By default, it uses a DecisionTreeClassifier with max_depth=1 (a decision stump)
adaboost_model = AdaBoostClassifier(
    estimator=DecisionTreeClassifier(max_depth=1),
    n_estimators=50,       # Number of sequential models to train
    learning_rate=1.0,     # Controls how aggressively weights are adjusted
    random_state=42
)

# 4. Train the model
adaboost_model.fit(X_train, y_train)

# 5. Make predictions
y_pred = adaboost_model.predict(X_test)

# 6. Print Evaluation Metrics
print(f"Accuracy Score: {accuracy_score(y_test, y_pred):.4f}\n")
print("Classification Report:")
print(classification_report(y_test, y_pred, target_names=iris.target_names))
```
