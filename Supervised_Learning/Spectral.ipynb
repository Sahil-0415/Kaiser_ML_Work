```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import SpectralClustering
from sklearn.datasets import make_moons

# Create sample data
X, y = make_moons(n_samples=300,
                  noise=0.05,
                  random_state=42)

# Apply Spectral Clustering
spectral = SpectralClustering(
    n_clusters=2,
    affinity='nearest_neighbors',
    n_neighbors=10,
    random_state=42
)

labels = spectral.fit_predict(X)

# Plot results
plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1],
            c=labels,
            cmap='viridis',
            s=50)

plt.title("Spectral Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

```python
from sklearn.datasets import make_moons
from sklearn.cluster import SpectralClustering, DBSCAN
import matplotlib.pyplot as plt

X, _ = make_moons(
    n_samples=300,
    noise=0.05,
    random_state=42
)

spectral_labels = SpectralClustering(
    n_clusters=2,
    affinity='nearest_neighbors'
).fit_predict(X)

dbscan_labels = DBSCAN(
    eps=0.2,
    min_samples=5
).fit_predict(X)

plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.scatter(X[:,0], X[:,1],
            c=spectral_labels,
            cmap='rainbow')
plt.title("Spectral Clustering")

plt.subplot(1,2,2)
plt.scatter(X[:,0], X[:,1],
            c=dbscan_labels,
            cmap='rainbow')
plt.title("DBSCAN")

plt.show()
```

```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_circles
from sklearn.cluster import SpectralClustering

# Generate ring-shaped dataset
X, y = make_circles(
    n_samples=500,
    factor=0.4,      # size of inner circle
    noise=0.03,
    random_state=42
)

# Apply Spectral Clustering
spectral = SpectralClustering(
    n_clusters=2,
    affinity='nearest_neighbors',
    n_neighbors=10,
    random_state=42
)

labels = spectral.fit_predict(X)

# Visualization
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.scatter(X[:,0], X[:,1], c=y, cmap='viridis')
plt.title("Actual Ring Clusters")

plt.subplot(1,2,2)
plt.scatter(X[:,0], X[:,1], c=labels, cmap='viridis')
plt.title("Spectral Clustering Result")

plt.show()
```
