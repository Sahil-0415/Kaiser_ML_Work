```python
data = [5, 7, 9, 10, 14, 15]

n = len(data)
print(n)
mean = sum(data) / n
# Sum of squared differences from the mean
sq_diffs = [(x - mean) ** 2 for x in data]

# Population variance
population_variance = sum(sq_diffs) / n

# Sample variance
sample_variance = sum(sq_diffs) / (n - 1)

print("Manual Population Variance:", round(population_variance,2))
print("Manual Sample Variance:", sample_variance)
```

```python
# The dataset
data = [85, 88, 92, 94, 96, 98, 100, 100, 100, 100]

# --- Step 1: Calculation of Mean ---
total_sum = sum(data)
num_elements = len(data)
mean_value = total_sum / num_elements

print("Step 1: Mean =", mean_value)

# --- Step 2: Calculation of Median ---
# Sort the data first: [85, 88, 92, 94, 96, 98, 100, 100, 100, 100]
sorted_data = sorted(data)

# Find the two middle numbers (5th and 6th items)
middle1 = sorted_data[4] # 96
middle2 = sorted_data[5] # 98
median_value = (middle1 + middle2) / 2

print("Step 2: Median =", median_value)

squared_diff_sum = 0
for x in data:
    squared_diff_sum = squared_diff_sum + (x - mean_value) ** 2

variance = squared_diff_sum / num_elements
exact_std_dev = variance ** 0.5  # ** 0.5 takes the square root

print("Step 3: Variance =", variance)
print("        Exact Standard Deviation =", exact_std_dev)

# Using your rounded value of 5 for step 5
std_dev_rounded = 5

mode_value = 100
print("Step 4: Mode =", mode_value)


sk_median = (3 * (mean_value - median_value)) / std_dev_rounded


sk_mode = (mean_value - mode_value) / std_dev_rounded

print("Step 5:")
print("   Skewness (Mean & Median) =", round(sk_median,3))
print("   Skewness (Mean & Mode) =", round(sk_mode,3))
```
