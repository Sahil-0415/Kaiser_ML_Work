```python
# Raw input data simulating a messy string with headers, footers, and extra spacing
raw_path = "   ###PREFIX_report_2026_final.pdf###   "

# 1. strip() -> Removes the wrapping spaces from both sides
clean_spaces = raw_path.strip()
print(f"1. After strip():       '{clean_spaces}'")

# 2. lstrip() -> Removes the leading '#' characters from the left
left_clean = clean_spaces.lstrip('#')
print(f"2. After lstrip():      '{left_clean}'")

# 3. rstrip() -> Removes the trailing '#' characters from the right
both_clean = left_clean.rstrip('#')
print(f"3. After rstrip():      '{both_clean}'")

# 4. removeprefix() -> Safely drops the exact starting keyword
no_prefix = both_clean.removeprefix("PREFIX_")
print(f"4. After removeprefix(): '{no_prefix}'")

# 5. removesuffix() -> Safely drops the exact ending file extension
final_output = no_prefix.removesuffix(".pdf")
print(f"5. Final Clean Name:    '{final_output}'")
```

```python
# The target URL to test
# url = "https://www.kaiser.com/index.html"
url = "https://www.kaiser.com"


print(f"Analyzing URL: {url}\n" + "="*35)

if len(url) <= 19 :
    print("⚠ The Length is Less Its a invalid website !!!!")

# 1. Check Protocol (SSL/Security)
if url.startswith("https://"):
    print("✓ Protocol: Secure (HTTPS)")
elif url.startswith("http://"):
    print("⚠ Protocol: Unsecure (HTTP)")
else:
    print("✗ Protocol: Invalid or missing protocol")

# 2. Check File Extension / Target Page
if url.endswith(".html"):
    print("✓ File Type: Standard HTML web page")
elif url.endswith(".php") or url.endswith(".asp"):
    print("✓ File Type: Dynamic server page")
else:
    print("⚠ File Type: No standard web extension found")

# 3. Check Specific Page Name
if "index.html" in url:
    print("✓ Destination: Points to the Homepage (index.html)")

# 4. Check for World Wide Web (www) subdomain
if "www." in url:
    print("✓ Subdomain: Contains 'www.'")

# 5. Extract and Validate Domain Name via Slicing & Find
# Find where the domain details actually start (after '//')
protocol_end = url.find("//") + 2 

# Find the next slash that separates the domain from the page path
path_start = url.find("/", protocol_end)

# Slice the domain substring out dynamically
domain = url[protocol_end:path_start]
print(f"✓ Extracted Domain: '{domain}'")

if "kaiser" in domain:
    print("✓ Brand Check: URL belongs to the 'kaiser' brand")
```
