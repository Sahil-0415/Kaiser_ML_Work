```python
# Create sample sets to use in the examples
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}
set_c = {1, 2}

print(f"Initial Sets:\nset_a = {set_a}\nset_b = {set_b}\nset_c = {set_c}\n" + "-"*40)

# ==========================================
# 1. ADDING & REMOVING ELEMENTS
# ==========================================
print("\n--- 1. Adding & Removing Elements ---")

# add(): Adds a single element
example_set = {1, 2}
example_set.add(3)
print(f"add(3): {example_set}")

# update(): Adds multiple elements from another collection
example_set.update([4, 5, 6])
print(f"update([4, 5, 6]): {example_set}")

# remove(): Removes an item (raises a KeyError if item doesn't exist)
example_set.remove(6)
print(f"remove(6): {example_set}")

# discard(): Removes an item safely (does NOT raise an error if missing)
example_set.discard(99) 
print(f"discard(99) [safe]: {example_set}")

# pop(): Removes and returns an arbitrary (random) element
popped_val = example_set.pop()
print(f"pop(): Removed '{popped_val}', Remaining set: {example_set}")

# clear(): Empties the entire set
clear_example = {1, 2, 3}
clear_example.clear()
print(f"clear(): {clear_example}")


# ==========================================
# 2. COMPARISONS & RELATIONSHIPS
# ==========================================
print("\n--- 2. Comparisons & Relationships ---")

# issubset(): Checks if all elements are contained in another set
print(f"set_c.issubset(set_a): {set_c.issubset(set_a)}")

# issuperset(): Checks if a set contains all elements of another set
print(f"set_a.issuperset(set_c): {set_a.issuperset(set_c)}")

# isdisjoint(): Checks if two sets have NO elements in common
set_d = {7, 8}
print(f"set_a.isdisjoint(set_d): {set_a.isdisjoint(set_d)}")


# ==========================================
# 3. SET MATHEMATICS (RETURNS NEW SET)
# ==========================================
print("\n--- 3. Set Mathematics (Returns New Set) ---")

# union(): Combines elements from both sets
print(f"union(): {set_a.union(set_b)}")

# intersection(): Gets elements common to both sets
print(f"intersection(): {set_a.intersection(set_b)}")

# difference(): Gets elements in set_a but NOT in set_b
print(f"difference(): {set_a.difference(set_b)}")

# symmetric_difference(): Gets elements in either set, but NOT both
print(f"symmetric_difference(): {set_a.symmetric_difference(set_b)}")

# copy(): Creates a shallow copy of the set
cloned_set = set_a.copy()
print(f"copy(): {cloned_set}")


# ==========================================
# 4. IN-PLACE SET MATHEMATICS (MODIFIES THE SET)
# ==========================================
print("\n--- 4. In-Place Set Mathematics (Modifies Original Set) ---")

# intersection_update(): Keeps ONLY items found in both sets
modify_set1 = {1, 2, 3}
modify_set1.intersection_update({2, 3, 4})
print(f"intersection_update(): {modify_set1}")

# difference_update(): Removes all items found in the other set
modify_set2 = {1, 2, 3}
modify_set2.difference_update({2, 3, 4})
print(f"difference_update(): {modify_set2}")

# symmetric_difference_update(): Keeps unique elements from both, removes duplicates
modify_set3 = {1, 2, 3}
modify_set3.symmetric_difference_update({2, 3, 4})
print(f"symmetric_difference_update(): {modify_set3}")
```
